```python
import pandas as pd
from kafka import KafkaProducer
import time
```

```python
df = pd.read_csv('.../.csv')  #add path to test.csv file 
# Select only the 'text' column
df = df['text']
```

```python
# Initialize Kafka Producer
producer = KafkaProducer(
    bootstrap_servers='localhost:9092',
    value_serializer=lambda v: v.encode('utf-8')  # Send value as plain string (no JSON)
)
count = 0
# Iterate through the rows in the DataFrame and send each tweet to Kafka as a string
for tweet in df:
    producer.send('test', value=tweet)
    print(f'Sent tweet: {tweet}')
    time.sleep(0.2)  
    count+=1
    if(count>20000):
        break
# Close the producer
producer.close()
```
